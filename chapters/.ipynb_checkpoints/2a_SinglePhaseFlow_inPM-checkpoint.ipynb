### Question 1 - Porosity and void ratio 1
The most fundamental of all porous material related properties is that of porosity $\phi$. It is defined as the ratio between volume of voids to that of the bulk material:

:::{math}
:label: por-eq-1
\begin{aligned}
\phi &= \frac{V_{void}}{V_{bulk}} \
V_{bulk} &= V_{pores}+V_{matrix}
\end{aligned}
:::
A related concept is that of the void ratio, which is more common in geotechnical applications.
:::{math}
:label: por-eq-2

e = \frac{V_{void}}{V_{matrix}}

:::

:::{tip} Solution Question 1
:class: dropdown

Use equations @por-eq-1 and @por-eq-2. For instance, start by writing expressions for $V_{\text{void}}$.

From equation @por-eq-1:
\begin{equation*}
V_{\text{void}} = \phi V_{\text{void}} + \phi V_{\text{matrix}}
\end{equation*}
From equation @por-eq-2:
\begin{equation*}
V_{\text{void}} = e V_{\text{matrix}}
\end{equation*}
Equate both expressions:
\begin{equation*}
\phi V_{\text{void}} + \phi V_{\text{matrix}} = e V_{\text{matrix}}
\end{equation*}
Divide by $V_{\text{matrix}}$:
\begin{equation*}
\phi \frac{V_{\text{void}}}{V_{\text{matrix}}} + \phi = e
\end{equation*}
Note that $\frac{V_{\text{void}}}{V_{\text{matrix}}}$ is the definition of $e$, so:
\begin{equation*}
\phi \cdot e + \phi = e
\end{equation*}
This can be reorganized to:
\begin{equation*}
\phi(1 + e) = e \implies \phi = \frac{e}{1 + e}
\end{equation*}
:::

#### a. Porosity ($\phi$)

* **Volume of solids ($V_s$):** 
The volume occupied by the solid matrix is calculated from the dry mass and the solid density:
\begin{equation*}
V_s = \frac{m_s}{\rho_s} = \frac{1318.3\text{ g}}{2.70\text{ g/cm}^3} \approx 488.26\text{ cm}^3
\end{equation*}
* **Volume of voids ($V_v$):** 
The total volume of the voids (pores filled with whatever fluid) is the difference between the total volume and the solid volume:
\begin{equation*}
V_v = V - V_s = 728.2\text{ cm}^3 - 488.26\text{ cm}^3 \approx 239.94\text{ cm}^3
\end{equation*}
* **Porosity ($\phi$):**
Porosity is defined as the ratio of void volume to the total volume:
\begin{equation*}
\phi = \frac{V_v}{V} = \frac{239.94\text{ cm}^3}{728.2\text{ cm}^3} \approx 0.3295
\end{equation*}
#### b. Void ratio ($e$)

The void ratio is defined as the ratio of void volume to the solid volume:
\begin{equation*}
e = \frac{V_v}{V_s} = \frac{239.94\text{ cm}^3}{488.26\text{ cm}^3} \approx 0.4914
\end{equation*}
#### c. Volumetric water content ($\theta$)
The volumetric water content is defined as the ratio of the volume of water ($V_w$) to the total volume ($V$).

* **Mass of water ($m_w$):** 

\begin{equation*}
m_w = m - m_s = 1523.6\text{ g} - 1318.3\text{ g} = 205.3\text{ g}
\end{equation*}
* **Volume of water ($V_w$):** *(Assuming $\rho_w = 1.0\text{ g/cm}^3$)*

\begin{equation*}
  V_w = \frac{m_w}{\rho_w} = 205.3\text{ cm}^3
\end{equation*}
* **Volumetric water content ($\theta$):**
\begin{equation*}
  \theta = \frac{V_w}{V} = \frac{205.3\text{ cm}^3}{728.2\text{ cm}^3} \approx 0.2819
\end{equation*}

#### d. Total saturated weight ($m_{\text{sat}}$)

When completely saturated, all the void space ($V_v$) is entirely filled with water.

* **Saturated mass of the lab sample ($m_{\text{sat}}$):**
\begin{equation*}
  m_{\text{sat}} = m_s + (V_v \times \rho_w) = 1318.3\text{ g} + (239.94\text{ cm}^3 \times 1.0\text{ g/cm}^3) = 1558.24\text{ g}
\end{equation*}
:::

#### a) Kozeny-Carman for a porous material with known matrix particle sizes

The Kozeny-Carman equation is

\begin{equation*}
k=\frac{d^{2}}{180}\frac{\phi^{3}}{(1-\phi)^{2}}
\end{equation*}

where:

- $k = \text{ intrinsic permeability } (\mathrm{m^2})$
- $d = \text{ representative grain diameter } (\mathrm{m})$
- $\phi = \text{ porosity}$

##### Step 1: Determine the effective grain size

The particle size distribution in @tab-psd gives:

| Size (mm) | Percent finer (%) |
|-----------|-------------------|
| 0.20 | 35 |
| 0.10 | 10 |

The effective diameter $d_{10}$ corresponds to 10% finer:

\begin{equation*}
d_{10}=0.10~\mathrm{ mm}
\end{equation*}

Convert to metres:

\begin{equation*}
d_{10}=0.10\times 10^{-3}
=1.0\times10^{-4}\ \mathrm{m}
\end{equation*}

##### Step 2: Calculate permeability for $\phi=0.37$

\begin{equation*}
k= \frac{(1.0\times10^{-4})^2}{180} \frac{(0.37)^3}{(1-0.37)^2}
\end{equation*}

\begin{equation*}
k= \frac{1.0\times10^{-8}}{180} \frac{0.050653}{0.3969}
\end{equation*}

\begin{equation*}
k=7.09\times10^{-12}\ \mathrm{m^2}
\end{equation*}

##### Answer

\begin{equation*}
\boxed{k=7.09\times10^{-12}\ \mathrm{m^2}}
\end{equation*}

#### b) Kozeny-Carman for a porous material with known matrix particle sizes with a porosity decrease

##### The new porosity is

\begin{equation*}
\phi_{\text{new}}
=0.37(1-0.25) = 0.2775
\end{equation*}

##### Using the same grain diameter:

\begin{equation*} 
k_{\text{new}} = \frac{(1.0\times10^{-4})^2}{180} \frac{(0.2775)^3}{(1-0.2775)^2}
\end{equation*}

\begin{equation*}
k_{\text{new}} = 5.56\times10^{-11}\left(\frac{0.021373}{0.522006}\right)
\end{equation*}

\begin{equation*}
k_{\text{new}}= 2.28 \times 10^{-12}\ \mathrm{m^2} 
\end{equation*}

##### Answer
\begin{equation*} 
\boxed{k_{\text{new}}=2.28\times10^{-12}\ \mathrm{m^2}} 
\end{equation*}

:::{warning}

##### Change in permeability
\begin{equation*} 
\frac{k_{\text{new}}}{k}= \frac{2.28\times10^{-12}} {7.09\times10^{-12}} = 0.321 
\end{equation*}

The intrinsic permeability is now only approximately one-third of its original value!
Note how the 25% reduction in porosity is leading to 67% permeability reduction.
:::

#### c) Hazen equation for clean sands (a particular porous material!)
The Hazen equation is an empirical relationship for estimating the hydraulic conductivity of clean sands:

\begin{equation*} 
K = C \cdot d_{10}^{2} 
\end{equation*}

where:

$K$ = hydraulic conductivity
$d_{10}$ = effective grain size
$C$ = empirical coefficient
For clean, uniformly graded sands, a commonly used value is

\begin{equation*} 
C = 100 
\end{equation*}

with $d_{10}$ expressed in cm and $K$ obtained in cm/s.

##### Step 1: Determine $d_{10}$
From @tab-psd:
| Size (mm) | Percent finer (%) |
|-----------|-------------------|
| 0.10      | 10                |

Therefore,

\begin{equation*} 
d_{10}=0.10\ \text{mm} 
\end{equation*}

Convert to cm:

\begin{equation*} 
d_{10}=0.01\ \text{cm} 
\end{equation*}

##### Step 2: Calculate hydraulic conductivity
\begin{equation*} 
K = 100(0.01)^2 
\end{equation*}

\begin{equation*} 
K = 0.01\ \text{cm/s} 
\end{equation*}

Convert to SI units:

\begin{equation*} 
K = 1.0\times10^{-4}\ \text{m/s} 
\end{equation*}

##### Step 3: Convert hydraulic conductivity to intrinsic permeability
The relationship between hydraulic conductivity and intrinsic permeability is

\begin{equation*} 
k=\frac{K\mu}{\rho g} 
\end{equation*}

Using properties of water at $20^\circ\text{C}$:

\begin{equation*} 
\mu = 1.002\times10^{-3}\ \text{Pa}\cdot\text{s} 
\end{equation*}

\begin{equation*} 
\rho = 1000\ \text{kg/m}^3 
\end{equation*}

\begin{equation*} 
g = 9.81\ \text{m/s}^2 
\end{equation*}

Substituting:

\begin{equation*} 
k\frac{(1.0\times10^{-4})(1.002\times10^{-3})} {(1000)(9.81)} 
\end{equation*}

\begin{equation*} 
k = 1.02\times10^{-11}\ \text{m}^2 
\end{equation*}

Answer
\begin{equation*} 
\boxed{k = 1.02\times10^{-11}\ \mathrm{m^2}} 
\end{equation*}

This estimate is somewhat larger than the Kozeny-Carman result because Hazen's equation is an empirical correlation developed for clean sands and tends to give relatively high permeability estimates.

### Question X - Hydraulic head 1
As you hopefully remember, the hydraulic head can be viewed as an expression of the Bernoulli equation, which in its turn is a form of conservation of energy. If you would check the dimensions of the Bernoulli equation, you will see that if you divide all terms in the equation by $\rho g$ the terms obtain the dimension of $[L]$. The benefit is that the somewhat abstract concept of energy is now expressed in a tanglible height. For instance, the height of the top of a water column above a chosen reference level. This is called the hydraulic head and its definition for slow flow, such as groundwater flow, is:
```{math}
:label: hh-eq-1

h \equiv \frac{p}{\rho g}+z

```

````python
display_quiz("../quiz/2_SPF_PM/q_Hydraulic_head_1.json")
````

:::{tip} Solution
:class: dropdown

#### Step 1: Calculate the exit velocity

The volumetric flow rate is

\begin{equation*}
Q = 2~\mathrm{L/s}
= 2\times10^{-3}~\mathrm{m^3/s}
\end{equation*}

The hose diameter is

\begin{equation*}
d = 2~\mathrm{cm}
= 0.02~\mathrm{m}
\end{equation*}

The cross-sectional area is

\begin{equation*}
A = \frac{\pi d^2}{4}
= \frac{\pi(0.02)^2}{4}
= 3.14\times10^{-4}~\mathrm{m^2}
\end{equation*}

The exit velocity is

\begin{equation*}
v = \frac{Q}{A}
= \frac{2\times10^{-3}}{3.14\times10^{-4}}
= 6.37~\mathrm{m/s}
\end{equation*}

#### Step 2: Determine the hydraulic head at the hose outlet

Bernoulli head is

\begin{equation*}
h = z + \frac{p}{\rho g} + \frac{v^2}{2g}
\end{equation*}

At the hose outlet, the water is exposed to the atmosphere, so

\begin{equation*}
\frac{p}{\rho g}=0
\end{equation*}

The elevation is

\begin{equation*}
z = 134.5~\mathrm{m}
\end{equation*}

The velocity head is:

\begin{equation*}
\frac{v^2}{2g} =\frac{(6.37)^2}{2(9.81)}=2.07~\mathrm{m}
\end{equation*}

Therefore,

\begin{equation*}
h = 134.5 + 2.07 = 136.57~\mathrm{m}
\end{equation*}

##### Answer: Hydraulic Head

\begin{equation*}
\boxed{h = 136.57~\mathrm{m}}
\end{equation*}

#### Step 3: Determine the maximum elevation reached by the water

Assuming the hydraulic head remains constant along the trajectory and that the velocity at the highest point is zero,

\begin{equation*}
v = 0
\end{equation*}

and the pressure is atmospheric:

\begin{equation*}
\frac{p}{\rho g}=0
\end{equation*}

Thus,

\begin{equation*}
h = z_{\max}
\end{equation*}

Substituting the previously calculated head:

\begin{equation*}
z_{\max}=136.57~\mathrm{m}
\end{equation*}

##### Answer: Maximum Elevation

The highest elevation reached by the water jet is

\begin{equation*}
\boxed{z_{\max}=136.57~\mathrm{m}}
\end{equation*}

This corresponds to a rise above the hose outlet of

\begin{equation*}
136.57 - 134.5=2.07~\mathrm{m}
\end{equation*}

or

\begin{equation*}
\boxed{2.07~\mathrm{m}}
\end{equation*}

above the end of the hose.

\end{equation*}

##### Answer

\begin{equation*}
\boxed{k = 8.27\times10^{-13}~\mathrm{m^2}}
\end{equation*}

#### e) Determine the conductivity for different fluids

For a given porous medium, the intrinsic permeability remains constant. Hydraulic conductivity depends on the fluid according to

\begin{equation*}
K=\frac{k\rho g}{\mu}
\end{equation*}

##### Water

Using the original properties:

\begin{equation*}
K_{\mathrm{water}}=8.10\times10^{-6}~\mathrm{m/s}
\end{equation*}

\begin{equation*}
K_{\mathrm{water}}=
0.70~\mathrm{m/day}
\end{equation*}

##### Leaded gasoline

Assume

\begin{equation*}
\rho = 740~\mathrm{kg/m^3}
\end{equation*}

\begin{equation*}
\mu = 6.0\times10^{-4}~\mathrm{Pa\,s}
\end{equation*}

Then

\begin{equation*}
K_{\mathrm{gasoline}}=\frac{(8.27\times10^{-13})(740)(9.81)}
{6.0\times10^{-4}}
\end{equation*}

\begin{equation*}
K_{\mathrm{gasoline}}=
1.00\times10^{-5}~\mathrm{m/s}
\end{equation*}

Converting to m/day:

\begin{equation*}
K_{\mathrm{gasoline}}
= 0.87~\mathrm{m/day}
\end{equation*}

##### Arabian medium crude oil

Assume

\begin{equation*}
\rho = 870~\mathrm{kg/m^3}
\end{equation*}

\begin{equation*}
\mu = 6.0\times10^{-3}~\mathrm{Pa\,s}
\end{equation*}

Then

\begin{equation*}
K_{\mathrm{crude}}
=\frac{(8.27\times10^{-13})(870)(9.81)}
{6.0\times10^{-3}}
\end{equation*}

\begin{equation*}
K_{\mathrm{crude}}
=1.18\times10^{-6}~\mathrm{m/s}
\end{equation*}

Converting to m/day:

\begin{equation*}
K_{\mathrm{crude}}
=0.10~\mathrm{m/day}
\end{equation*}

##### Answer
The hydraulic conductivities are approximately

\begin{equation*}
\boxed{K_{\mathrm{water}} = 0.70~\mathrm{m/day}}
\end{equation*}

\begin{equation*}
\boxed{K_{\mathrm{gasoline}} = 0.87~\mathrm{m/day}}
\end{equation*}

\begin{equation*}
\boxed{K_{\mathrm{crude}} = 0.10~\mathrm{m/day}}
\end{equation*}

The conductivity is highest for gasoline because of its relatively low viscosity and lowest for crude oil because of its much higher viscosity.

### Question X - Falling head test

````python
display_quiz("../quiz/2_SPF_PM/q_Falling_head.json")
````

::::{hint} Hint Question X
:class: dropdown
Derive the following equation: $\frac{dh}{dt} = -\frac{KA}{aL}h$. Can you do that? After you found out how to properly do this, you can solve this as a simple first order, linear ODE. Note $a$ is the diameter of the burette.
:::{warning}
There is no point just continuing without being able to derive the equation first. You need to understand its origin.
:::
::::    

:::{tip} Solution Question X
:class: dropdown

#### Step 1: Write the falling-head equation

For a falling-head permeameter,

\begin{equation*}
K=\frac{aL}{At}\ln\left(\frac{h_1}{h_2}\right)
\end{equation*}

Rearranging to solve for time:

\begin{equation*}
t=\frac{aL}{AK}\ln\left(\frac{h_1}{h_2}\right)
\end{equation*}

where:

- $a$ = cross-sectional area of the burette
- $A$ = cross-sectional area of the sample
- $L$ = sample length
- $K$ = hydraulic conductivity
- $h_1$ = initial head difference
- $h_2$ = final head difference

#### Step 2: Calculate the cross-sectional areas

The sample diameter is

\begin{equation*}
D=8~\mathrm{cm}
\end{equation*}

Therefore,

\begin{equation*}
A=\frac{\pi D^2}{4}
=\frac{\pi(8)^2}{4}
=50.27~\mathrm{cm^2}
\end{equation*}

The burette diameter is

\begin{equation*}
d=2~\mathrm{mm}
=0.2~\mathrm{cm}
\end{equation*}

Therefore,

\begin{equation*}
a=\frac{\pi d^2}{4}
=\frac{\pi(0.2)^2}{4}
=0.03142~\mathrm{cm^2}
\end{equation*}

#### Step 3: Substitute the known values

The hydraulic conductivity is

\begin{equation*}
K=10^{-4}~\mathrm{cm/s}
\end{equation*}

The sample length is

\begin{equation*}
L=22~\mathrm{cm}
\end{equation*}

The head changes from

\begin{equation*}
h_1=80~\mathrm{cm}
\end{equation*}

to

\begin{equation*}
h_2=20~\mathrm{cm}
\end{equation*}

Thus,

\begin{equation*}
\ln\left(\frac{h_1}{h_2}\right)=\ln\left(\frac{80}{20}\right)=\ln(4)=1.3863
\end{equation*}

Substituting into the falling-head equation:

\begin{equation*}
t=\frac{(0.03142)(22)}{(50.27)(10^{-4})}\times 1.3863
\end{equation*}

\begin{equation*}
t=190.8~\mathrm{s}
\end{equation*}

#### Answer

\begin{equation*}
\boxed{t \approx 191~\mathrm{s}}
\end{equation*}

or approximately

\begin{equation*}
\boxed{3.18~\mathrm{minutes}}
\end{equation*}
