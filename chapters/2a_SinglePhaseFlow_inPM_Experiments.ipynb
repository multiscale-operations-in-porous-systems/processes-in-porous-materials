:::{tip} Solution Question 10
:class: dropdown
***a) Hydraulic conductivity***
**Step 1: Determine the hydraulic gradient**

The head difference across the sample is

\begin{equation*}
\Delta h = h_1-h_2
\end{equation*}

Substituting the given values:

\begin{equation*}
\Delta h = 95.0-37.0
=58.0~\mathrm{cm}
\end{equation*}

The hydraulic gradient is

\begin{equation*}
i=\frac{\Delta h}{L}
\end{equation*}

\begin{equation*}
i=\frac{58.0}{20.0}
=2.90
\end{equation*}

**Step 2: Calculate the cross-sectional area**

The sample radius is

\begin{equation*}
r=1.5~\mathrm{cm}
\end{equation*}

The cross-sectional area is

\begin{equation*}
A=\pi r^2
\end{equation*}

\begin{equation*}
A=\pi(1.5)^2
=7.069~\mathrm{cm^2}
\end{equation*}

**Step 3: Calculate the specific discharge**

The volumetric flow rate is

\begin{equation*}
Q=3.5~\mathrm{cm^3/min}
\end{equation*}

Convert to \(\mathrm{cm^3/s}\):

\begin{equation*}
Q=\frac{3.5}{60}
=0.05833~\mathrm{cm^3/s}
\end{equation*}

The specific discharge is

\begin{equation*}
q=\frac{Q}{A}
\end{equation*}

\begin{equation*}
q = \frac{0.05833}{7.069} = 0.00825~\mathrm{cm/s}
\end{equation*}

**Step 4: Calculate the hydraulic conductivity**

Darcy's law gives

\begin{equation*}
q=Ki
\end{equation*}

Therefore,

\begin{equation*}
K=\frac{q}{i}
\end{equation*}

\begin{equation*}
K=\frac{0.00825}{2.90}=2.85\times10^{-3}~\mathrm{cm/s}
\end{equation*}

**Answer**

\begin{equation*}
\boxed{K=2.85\times10^{-3}~\mathrm{cm/s}}
\end{equation*}

***c) Calculate the average linear velocity***

Darcy's law gives the specific discharge:

\begin{equation*}
q_x = K i
\end{equation*}

Substituting the given values:

\begin{equation*}
q_x = (0.7)(0.012)
= 0.0084~\mathrm{m/day}
\end{equation*}

The average linear velocity is related to the specific discharge by

\begin{equation*}
\overline{v}_x=\frac{q_x}{\phi}
\end{equation*}

Therefore,

\begin{equation*}
\overline{v}_x =
\frac{0.0084}{0.32} =
0.02625~\mathrm{m/day}
\end{equation*}

**Answer**

\begin{equation*}
\boxed{\overline{v}_x = 0.03~\mathrm{m/day}}
\end{equation*}

***d) Calculate the intrinsic permeability***

The relationship between hydraulic conductivity and intrinsic permeability is

\begin{equation*}
K=\frac{k\rho g}{\mu}
\end{equation*}

Rearranging:

\begin{equation*}
k=\frac{K\mu}{\rho g}
\end{equation*}

First convert the hydraulic conductivity to SI units:

\begin{equation*}
K=0.7~\mathrm{m/day}=
\frac{0.7}{86400}=
8.10\times10^{-6}~\mathrm{m/s}
\end{equation*}

For water at \(20^\circ\mathrm{C}\),

\begin{equation*}
\mu = 1.002\times10^{-3}~\mathrm{Pa\,s}
\end{equation*}

\begin{equation*}
\rho = 1000~\mathrm{kg/m^3}
\end{equation*}

\begin{equation*}
g = 9.81~\mathrm{m/s^2}
\end{equation*}

Substituting:

\begin{equation*}
k=\frac{(8.10\times10^{-6})(1.002\times10^{-3})}
{(1000)(9.81)}
\end{equation*}

\begin{equation*}
k = 8.27\times10^{-13}~\mathrm{m^2}
\end{equation*}

**Answer**

\begin{equation*}
\boxed{k = 8.27\times10^{-13}~\mathrm{m^2}}
\end{equation*}

***e) Determine the conductivity for different fluids***

For a given porous medium, the intrinsic permeability remains constant. Hydraulic conductivity depends on the fluid according to

\begin{equation*}
K=\frac{k\rho g}{\mu}
\end{equation*}

**Water**

Using the original properties:

\begin{equation*}
K_{\mathrm{water}}=8.10\times10^{-6}~\mathrm{m/s}
\end{equation*}

\begin{equation*}
K_{\mathrm{water}}=
0.70~\mathrm{m/day}
\end{equation*}

**Leaded gasoline**

Assume

\begin{equation*}
\rho = 740~\mathrm{kg/m^3}
\end{equation*}

\begin{equation*}
\mu = 6.0\times10^{-4}~\mathrm{Pa\,s}
\end{equation*}

Then

\begin{equation*}
K_{\mathrm{gasoline}}=\frac{(8.27\times10^{-13})(740)(9.81)}
{6.0\times10^{-4}}
\end{equation*}

\begin{equation*}
K_{\mathrm{gasoline}}=
1.00\times10^{-5}~\mathrm{m/s}
\end{equation*}

Converting to m/day:

\begin{equation*}
K_{\mathrm{gasoline}}
= 0.87~\mathrm{m/day}
\end{equation*}

**Arabian medium crude oil**

Assume

\begin{equation*}
\rho = 870~\mathrm{kg/m^3}
\end{equation*}

\begin{equation*}
\mu = 6.0\times10^{-3}~\mathrm{Pa\,s}
\end{equation*}

Then

\begin{equation*}
K_{\mathrm{crude}}
=\frac{(8.27\times10^{-13})(870)(9.81)}
{6.0\times10^{-3}}
\end{equation*}

\begin{equation*}
K_{\mathrm{crude}}
=1.18\times10^{-6}~\mathrm{m/s}
\end{equation*}

Converting to m/day:

\begin{equation*}
K_{\mathrm{crude}}
=0.10~\mathrm{m/day}
\end{equation*}

**Answer**
The hydraulic conductivities are approximately

\begin{equation*}
\boxed{K_{\mathrm{water}} = 0.70~\mathrm{m/day}}
\end{equation*}

\begin{equation*}
\boxed{K_{\mathrm{gasoline}} = 0.87~\mathrm{m/day}}
\end{equation*}

\begin{equation*}
\boxed{K_{\mathrm{crude}} = 0.10~\mathrm{m/day}}
\end{equation*}

The conductivity is highest for gasoline because of its relatively low viscosity and lowest for crude oil because of its much higher viscosity.

### Question 11 - Falling head test

````python
display_quiz("../quiz/2_SPF_PM/q_Falling_head.json")
````

::::{hint} Hint Question 11
:class: dropdown
Derive the following equation: $\frac{dh}{dt} = -\frac{KA}{aL}h$. Can you do that? After you found out how to properly do this, you can solve this as a simple first order, linear ODE. Note $a$ is the diameter of the burette.
:::{warning}
There is no point just continuing without being able to derive the equation first. You need to understand its origin.
:::
::::    

:::{tip} Solution Question 11
:class: dropdown

***Step 1: Write the falling-head equation***

For a falling-head permeameter,

\begin{equation*}
K=\frac{aL}{At}\ln\left(\frac{h_1}{h_2}\right)
\end{equation*}

Rearranging to solve for time:

\begin{equation*}
t=\frac{aL}{AK}\ln\left(\frac{h_1}{h_2}\right)
\end{equation*}

where:

- $a$ = cross-sectional area of the burette
- $A$ = cross-sectional area of the sample
- $L$ = sample length
- $K$ = hydraulic conductivity
- $h_1$ = initial head difference
- $h_2$ = final head difference

***Step 2: Calculate the cross-sectional areas***

The sample diameter is

\begin{equation*}
D=8~\mathrm{cm}
\end{equation*}

Therefore,

\begin{equation*}
A=\frac{\pi D^2}{4}
=\frac{\pi(8)^2}{4}
=50.27~\mathrm{cm^2}
\end{equation*}

The burette diameter is

\begin{equation*}
d=2~\mathrm{mm}
=0.2~\mathrm{cm}
\end{equation*}

Therefore,

\begin{equation*}
a=\frac{\pi d^2}{4}
=\frac{\pi(0.2)^2}{4}
=0.03142~\mathrm{cm^2}
\end{equation*}

***Step 3: Substitute the known values***

The hydraulic conductivity is

\begin{equation*}
K=10^{-4}~\mathrm{cm/s}
\end{equation*}

The sample length is

\begin{equation*}
L=22~\mathrm{cm}
\end{equation*}

The head changes from

\begin{equation*}
h_1=80~\mathrm{cm}
\end{equation*}

to

\begin{equation*}
h_2=20~\mathrm{cm}
\end{equation*}

Thus,

\begin{equation*}
\ln\left(\frac{h_1}{h_2}\right)=\ln\left(\frac{80}{20}\right)=\ln(4)=1.3863
\end{equation*}

Substituting into the falling-head equation:

\begin{equation*}
t=\frac{(0.03142)(22)}{(50.27)(10^{-4})}\times 1.3863
\end{equation*}

\begin{equation*}
t=190.8~\mathrm{s}
\end{equation*}

**Answer**

\begin{equation*}
\boxed{t \approx 191~\mathrm{s}}
\end{equation*}

or approximately

\begin{equation*}
\boxed{3.18~\mathrm{minutes}}
\end{equation*}
