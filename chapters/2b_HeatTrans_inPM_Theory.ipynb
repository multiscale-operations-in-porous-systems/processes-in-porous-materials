Let $V_{total}$ be a unit volume of the porous medium.
The solid volume fraction is $\frac{V_s}{V_{total}} =1−\phi=1−0.4=0.6$.

For a single solid cylinder of length $L$ and radius $R$ ($d_p=2R$):
Volume of one cylinder: 

$V_{cyl} = \pi R^2 L$

Lateral surface area of one cylinder: 

$A_{cyl} =2\pi RL$

Find the ratio of surface area to solid volume for a cylinder: 

$\frac{A_{cyl}}{V_{cyl}} =\frac{2\pi RL}{\pi R^2 L}=\frac{2}{R}=\frac{4}{d_p}$ 

The specific interfacial area $a_{fs}$ is the total surface area divided by the total volume $V_{total}$): 

$a_{sf}= \frac{A_{total}}{V_{total}} = \frac{A_{total}}{V_s} \times \frac{V_s}{V_{total}}  = (\frac{4}{d_p} \times (1−\phi))$

Plug in the numbers ($d_p=0.002$ m, $\phi=0.4$): $a_{sf}= 4/(0.002 ~m) \times (1−0.4)=2000 \times 0.6= 1200~\text{m}^2/\text{m}^3$

:::{tip} Solution Question 4
:class: dropdown

(a) Pressure Drop ($\Delta P$): 
According to Darcy's Law (which you already know): $\vec{v}_0=−\frac{K}{\mu} \nabla P \rightarrow \Delta P \propto v_0 $ 
Because velocity $v_0$ doubles ($2v_0$), the pressure drop doubles ($2\Delta P$).

(b) Heat Transfer Coefficient ($h$):
The porous Péclet number is directly proportional to velocity: $Pe_p \propto v_0$. If velocity doubles, $Pe_p$ doubles. Given the empirical correlation $Nu_p \propto Pe_p ^{0.5}$, we know that:$Nu_p=\frac{h d_p}{k_f} \propto (2Pe_p)^{0.5}=\sqrt{2}Pe^{0.5}$ Therefore, the convective heat transfer coefficient increases by a factor of $\sqrt{2}=1.41$ (a 41% increase).

The lesson here is that doubling the flow gives you 41% better cooling, but costs you 100% more in pressure penalty (and even higher pumping power since Power $\propto v_0 \cdot \Delta P \propto v_0^2$). This showcases a classical engineering optimization challenge.
:::

Exercise 6: The Flash Method Experiment (Transient Conduction)

Problem: In a standard laser flash experiment, a thin porous sample of thickness L = 4 mm and porosity φ = 0.5 is saturated with static water. A short pulse of laser energy is shot at the front face (x=0), and the temperature rise Δ T(t) is monitored at the back face (x=L). The half-rise time (the time it takes for the back face to reach half of its maximum temperature rise) is measured to be \(t_{1/2} = 5.37\text{ s}\).
Assuming Parker's classical relation holds (\(t_{1/2} = 0.1388 \cdot L^2 / \alpha_{\text{eff}}\)), calculate the Effective Thermal Conductivity (\(k_{\text{eff}}\)) under Local Thermal Equilibrium (LTE).
• Properties: \(\rho_f = 1000\text{ kg/m}^3\), \(c_{pf} = 4200\text{ J/kgK}\), \(\rho_s = 2500\text{ kg/m}^3\), \(c_{ps} = 800\text{ J/kgK}\).
💡 Hint:
1. Use Parker's relation to isolate and calculate the effective thermal diffusivity (\(\alpha _{\text{eff}}\)). Don't forget to convert the sample thickness L to meters before squaring it.
2. Under LTE, the effective volumetric heat capacity \((\rho c_p)_{\text{eff}}\) is a basic volume-weighted average of both the fluid and solid phases based on the porosity φ.
3. Link conductivity to diffusivity using the fundamental identity: \(k_{\text{eff}} = \alpha_{\text{eff}} \cdot (\rho c_p)_{\text{eff}}\).
✔️ Solution:
1. Calculate Effective Thermal Diffusivity (\(\alpha _{\text{eff}}\)):
\(\alpha _{\text{eff}}=\frac{0.1388\cdot L^{2}}{t_{1/2}}=\frac{0.1388\cdot (0.004\text{\ m})^{2}}{5.37\text{\ s}}=4.136\times 10^{-7}\text{\ m}^{2}/\text{s}\)
2. Calculate Effective Volumetric Heat Capacity \((\rho c_p)_{\text{eff}}\):
\((\rho c_{p})_{\text{eff}}=\phi (\rho c_{p})_{f}+(1-\phi )(\rho c_{p})_{s}\)
\((\rho c_{p})_{\text{eff}}=0.5\cdot (1000\cdot 4200)+0.5\cdot (2500\cdot 800)\)
\((\rho c_{p})_{\text{eff}}=2.1\times 10^{6}+1.0\times 10^{6}=3.1\times 10^{6}\text{\ J/m}^{3}\text{K}\)
3. Calculate Effective Thermal Conductivity (\(k_{\text{eff}}\)):
\(k_{\text{eff}}=\alpha _{\text{eff}}\cdot (\rho c_{p})_{\text{eff}}=(4.136\times 10^{-7}\text{\ m}^{2}/\text{s})\cdot (3.1\times 10^{6}\text{\ J/m}^{3}\text{K})\approx \mathbf{1.282}\text{\ W/m·K}\)

Exercise 7: The Schumann Column Experiment (LTNE Characterization)

Problem: In an idealized Schumann packed bed experiment, cold liquid is pumped at a high, constant velocity through a hot porous matrix to evaluate the inter-phase heat transfer coefficient (\(h_{sf}\)). At a specific monitoring probe inside the bed, sensors register that the solid phase temperature stays at \(T_s = 80^\circ\text{C}\) while the fluid moving past it tracks at \(T_f = 40^\circ\text{C}\). The volumetric interfacial surface area is \(a_{sf} = 2000\text{ m}^2/\text{m}^3\). If the local steady heat transfer rate per unit volume absorbed by the fluid is quantified as q''' = 4.0 × 10⁵ W/m³, determine the value of \(h_{sf}\).
💡 Hint:
1. Think about the source term in the Local Thermal Non-Equilibrium (LTNE) energy equations. The thermal energy exchange between the fluid and solid phases per unit volume is proportional to the local temperature difference.
2. Set up Newton's law of cooling modified for volume averaging: \(q''' = h_{sf} a_{sf} (T_s - T_f)\).
3. Isolate \(h_{sf}\) and solve.
✔️ Solution:
1. Identify the LTNE Energy Exchange Equation:
\(q^{\prime \prime \prime }=h_{sf}a_{sf}(T_{s}-T_{f})\)
2. Rearrange for \(h_{sf}\):
\(h_{sf}=\frac{q^{\prime \prime \prime }}{a_{sf}(T_{s}-T_{f})}\)
3. Substitute values:
\(h_{sf}=\frac{4.0\times 10^{5}\text{\ W/m}^{3}}{2000\text{\ m}^{2}/\text{m}^{3}\cdot (80^{\circ }\text{C}-40^{\circ }\text{C})}=\frac{400,000}{2000\cdot 40}=\frac{400,000}{80,000}=\mathbf{5.0}\text{\ W/m}^{\mathbf{2}}\text{K}\)

Exercise 8: Transverse Dispersion Measurement (Tracer Breakthrough)

Problem: A thermal tracer experiment is conducted to track hydro-thermal mixing across a flat porous slab. Water flows steadily through the bed along the x-axis. A line heat source acts continuously at the inlet. The total effective thermal conductivity in the lateral y-direction (\(k_{y,\text{total}}\)) is monitored over varying velocities.
When the flow velocity is doubled (2 × v₀), the experimentalists observe that the static thermal conductivity contribution (\(k_{e}\)) remains completely unchanged, yet the overall thermal conductivity in the y-direction (\(k_{y,\text{total}}\)) increases by exactly 60%. What percentage of the initial total lateral heat transport was driven by thermal dispersion?
💡 Hint:
1. Write out the total lateral conductivity equation as the sum of its static and dynamic parts: \(k_{y,\text{total}} = k_e + k_{\text{disp}}\).
2. Recall how dispersion scales with velocity in typical pore regimes: \(k_{\text{disp}}\) is directly proportional to velocity (\(k_{\text{disp}} \propto v_0\)). Therefore, if velocity doubles, \(k_{\text{disp}}\) doubles.
3. Set up an algebraic equation where the new total conductivity equals 1.6 times the initial total conductivity, then solve for the ratio of \(k_{\text{disp}}\) to \(k_{e}\).
✔️ Solution:
1. Establish the Scaling System:
	• Initial condition: \(k_{y,\text{total, initial}} = k_e + k_{\text{disp}}\)
	• New condition (velocity doubled): \(k_{y,\text{total, new}} = k_e + 2 \cdot k_{\text{disp}}\)
2. Apply the 60% Increase Constraint:
\(k_{e}+2\cdot k_{\text{disp}}=1.6\cdot (k_{e}+k_{\text{disp}})\)
\(k_{e}+2\cdot k_{\text{disp}}=1.6\cdot k_{e}+1.6\cdot k_{\text{disp}}\)
3. Isolate the Variables:
\(2\cdot k_{\text{disp}}-1.6\cdot k_{\text{disp}}=1.6\cdot k_{e}-k_{e}\)
\(0.4\cdot k_{\text{disp}}=0.6\cdot k_{e}\implies k_{\text{disp}}=1.5\cdot k_{e}\)
4. Calculate Initial Dispersion Percentage Contribution:
\(\text{Share}=\frac{k_{\text{disp}}}{k_{e}+k_{\text{disp}}}=\frac{1.5\cdot k_{e}}{k_{e}+1.5\cdot k_{e}}=\frac{1.5}{2.5}=0.60=\mathbf{60\%}\)

Exercise 9: Horton-Rogers-Lapwood Apparatus (Onset of Convection)

Problem: Students are validating natural convection limits using an adjustable horizontal porous layer bounded by isothermal copper sheets. The cavity contains a porous matrix filled with silicone oil. The temperature difference is increased until convection loops initiate.
In Run A, with a layer height of H = 10 cm, the onset of natural convection occurs at exactly \(\Delta T_{\text{crit}} = 12^\circ\text{C}\). If they replace the setup with an identical porous material but decrease the height to H = 5 cm (Run B), what will the new \(\Delta T_{\text{crit}}\) be?
💡 Hint:
1. Natural convection in a porous cell triggers at a specific, fixed threshold called the critical Porous Rayleigh number (\(Ra_{p,\text{crit}} = 4\pi^2\)).
2. Review the definition of the Porous Rayleigh number: \(Ra_p = \frac{\rho g \beta K \Delta T H}{\mu \alpha_{\text{eff}}}\). Notice how it scales with layer height H. Unlike clear fluids which scale with H³, the porous equivalent scales linearly with H due to Darcy resistance.
3. Since all fluid and material parameters are constant, the product \((\Delta T_{\text{crit}} \cdot H)\) must remain constant to trigger convection.
✔️ Solution:
1. Identify Constant Scaling Relations:
Because all fluid and structural parameters remain unchanged, the system is governed by:
\(Ra_{p,\text{crit}}\propto \Delta T_{\text{crit}}\cdot H=\text{constant}\)
2. Equate Run A and Run B:
\(\Delta T_{\text{crit,\ A}}\cdot H_{A}=\Delta T_{\text{crit,\ B}}\cdot H_{B}\)
3. Solve for the New Threshold (\(\Delta T_{\text{crit, B}}\)):
\(\Delta T_{\text{crit,\ B}}=\Delta T_{\text{crit,\ A}}\cdot \left(\frac{H_{A}}{H_{B}}\right)\)
\(\Delta T_{\text{crit,\ B}}=12^{\circ }\text{C}\cdot \left(\frac{10\text{\ cm}}{5\text{\ cm}}\right)=12\cdot 2=\mathbf{24}^{\mathbf{\circ }}\text{C}\)

Exercise 10: Conjugate Pore-Scale Boundary Layer Scaling

Problem: In a localized micro-CT validated fluid channel experiment within a highly porous metal structure, the local pore-scale boundary layer thickness \(\delta _{T}\) is analyzed. For a baseline flow condition with a pore Péclet number of \(Pe_p = 16\), the average thermal boundary layer thickness inside a pore space channels out to \(\delta_T = 150\,\mu\text{m}\). If the external pump system steps up the flow rate so that the Péclet number shifts to \(Pe_p = 64\), estimate the new boundary layer thickness \(\delta _{T}\) assuming standard boundary layer scaling (\(\delta_T \propto Pe_p^{-0.5}\)).
💡 Hint:
1. Set up a proportional scaling ratio comparing the new state to the initial state.
2. The problem states that the scaling matches \(\delta_T \propto Pe_p^{-0.5}\), which can be rewritten as \(\delta_T \propto \frac{1}{\sqrt{Pe_p}}\).
3. Compute the ratio of the new Péclet number to the old one, evaluate the square root, and adjust the initial boundary layer thickness accordingly.
✔️ Solution:
1. Formulate the Proportionality Ratio:
\(\frac{\delta _{T,\text{new}}}{\delta _{T,\text{initial}}}=\left(\frac{Pe_{p,\text{new}}}{Pe_{p,\text{initial}}}\right)^{-0.5}\)
2. Substitute Péclet Values:
\(\frac{\delta _{T,\text{new}}}{150\,\mu \text{m}}=\left(\frac{64}{16}\right)^{-0.5}=(4)^{-0.5}=\frac{1}{\sqrt{4}}=0.5\)
3. Calculate Final Thickness:
\(\delta _{T,\text{new}}=150\,\mu \text{m}\cdot 0.5=\mathbf{75}\,\mathbf{\mu }\text{m}\)
