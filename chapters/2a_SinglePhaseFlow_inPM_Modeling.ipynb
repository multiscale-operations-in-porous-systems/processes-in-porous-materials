## Question 12 - Bundle of Tubes

A BTM is the easiest way to represent a porous material. Whether it is a good fit, depends on the goal you set and the material structure. The idea is that each pore can be represented with a capillary of a certain diameter and length. By considering not one pore, but N pores, an approximation of the whole porous material is obtained. We can start with only one fraction of pore characteristics and expand it to match a pore size distribution. In the end it is about matching the material's resistance to flow, i.e. find the appropriate expression for the intrinsic permeability. There are many permeability models that try to do this by looking at geometrical characteristics. One of them is the Kozeny-Carman model.

***Part 1: Intrinsic Permeability Calculation***
We are given a parallel Bundle of Tubes model where the total void space ($\phi = 0.30$) is split equally between two pore radii fractions ($f_1 = 0.5$ and $f_2 = 0.5$):
$$R_1 = 10 \times 10^{-6} \text{ m}$$
$$R_2 = 20 \times 10^{-6} \text{ m}$$

The governing equation derived from the Hagen-Poiseuille relationship for intrinsic permeability ($k$) of a parallel bundle of cylinders is:
$$k = \frac{\phi}{8} \sum (f_i R_i^2)$$

Substitute the known values into the summation:
$$\sum (f_i R_i^2) = 0.5 \cdot (10 \times 10^{-6} \text{ m})^2 + 0.5 \cdot (20 \times 10^{-6} \text{ m})^2$$
$$\sum (f_i R_i^2) = 0.5 \cdot (100 \times 10^{-12} \text{ m}^2) + 0.5 \cdot (400 \times 10^{-12} \text{ m}^2)$$
$$\sum (f_i R_i^2) = (50 \times 10^{-12}) + (200 \times 10^{-12}) = 250 \times 10^{-12} \text{ m}^2$$

Now, multiply by the porosity pre-factor:
$$k = \frac{0.30}{8} \cdot (250 \times 10^{-12} \text{ m}^2)$$
$$k = 0.0375 \cdot (250 \times 10^{-12} \text{ m}^2) = 9.375 \times 10^{-12} \text{ m}^2$$

***Part 2: Underlying Assumptions of Kozeny-Carman***
The classic Kozeny-Carman equation simplifies highly complex 3D porous structures into an equivalent hydraulic system. The core assumptions required to map a static, volumetric distribution directly to a dynamic permeability constant include:

* **Non-Interacting Parallel Channels:** The system is treated as a continuous bundle of channels with a uniform mean hydraulic radius. It assumes the fluid moves evenly through these channels without crossing over or stalling.
* **Neglecting Network Topology:** It ignores spatial bottlenecks (alternating expansions and contractions) and multi-directional routing loops (connectivity/coordination numbers).
* **Laminar Viscous Flow:** It assumes creeping flow conditions ($\text{Re} < 1$), where Darcy's Law holds valid and inertial effects are non-existent.
* **Isotropic Homogeneity:** The macroscopic properties are assumed to be evenly averaged across the sample continuum.

## Question 13 - Pore Network Model, Permeability (EXTRA)
A more advanced way of modeling a porous materials (and processes within) is pore network modelling. In this methods, the void space is divided in pore bodies and pore throats - often visualized as a 'stick-and-ball' structure. A great benefit is that these models tend to be computationally light. There are several types of pore network model, classified based on topology (structured or unstructured) and geometry/property (regular or irregular). The spatial arrangement and connection of pore bodies and throats to one another is tied to topology, whereas the uniformity or statistical distributions of the size, shape, length and coordination number are related to the geometry. This leads to four types: structured regular, structured irregular, unstructured regular and unstructured irregular.
Once the network topology and geometry are defined, it can be generated.

There are various experimental techniques that can tell you about the pore network in a porous material. Think about (3D) imaging techniques, such as (confocal) microscopy, micro-CT, AFM, but also more macroscopic techniques such as MICP measurements (yes, you can google that!). From those techniques statistics and geometries for pore bodies and throats can be derived.

After the lattice is in place, a pressure gradient can be applied on it. Based on rules wether a pore body or throat can be invaded, the network will fill up if initially empty. This is an example of 2-phase flow use. One can add other physics to it, such as heat transfer and multi component transport. For single phase flow, the main idea is to use that the whole system is full of e.g. water as an initial condition and apply a pressure gradient at designated inlet and outlet pores. That way, the local velocities inside de pores can be visualized an a permeability and hydraulic conductivity value can be obtained and compared to e.g. Darcy Constant Head tests.
