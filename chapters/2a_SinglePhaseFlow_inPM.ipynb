## Theory
### Porosity and void ratio
The most fundamental of all porous material related properties is that of porosity $\phi$. It is defined as the ratio between volume of voids to that of the bulk material:

```{math}
:label: por-eq-1
\begin{aligned}
\phi &= \frac{V_{void}}{V_{bulk}} \\
V_{bulk} &= V_{pores}+V_{matrix}
\end{aligned}
```
A related concept is that of the void ratio, which is more common in geotechnical applications.
```{math}
:label: por-eq-2

e = \frac{V_{void}}{V_{matrix}}

```

### Darcy's law
**Hydraulic head**<br>
As you hopefully remember, the hydraulic head can be viewed as an expression of the Bernoulli equation, which in its turn is a form of conservation of energy. If you would check the dimensions of the Bernoulli equation, you will see that if you divide all terms in the equation by $\rho g$ the terms obtain the dimension of $[L]$. The benefit is that the somewhat abstract concept of energy is now expressed in a tanglible height. For instance, the height of the top of a water column above a chosen reference level. This is called the hydraulic head and its definition for slow flow, such as groundwater flow, is:
```{math}
:label: hh-eq-1

h \equiv \frac{p}{\rho g}+z

```

`````{admonition} Elongation of a rod
:class: examp

A tensile force of 40000 N is applied to an aluminum rod with a diameter of 1.0 cm and a length of 2.0 meters. How far does the bar stretch? How much does the bar become thinner? See {numref}`tbl-charc-materials` for material data.

````{admonition} Solution
:class: sol dropdown

The following applies to the tension in the rod:

$$ \sigma_n = \frac{F}{A_0} = \frac{F}{\frac{1}{4}\pi d_0^2} = \frac{40000}{\frac{1}{4}\pi (0.01)^2} = 5.09 \cdot 10^8 \textrm{  Pa} $$ (mech-eq-6)

With the modulus of elasticity of aluminum of 70 GPa the elongation becomes:

$$ e = \frac{\sigma_n}{E} = \frac{5.09 \cdot 10^8}{70 \cdot 10^9} = 7.28 \cdot 10^{-3} \textrm{ [-]}$$ (mech-eq-7)

For a rod of $2.0 \textrm{ m}$ this gives an extension of: 
$\Delta L = L_0 \cdot e = 2.0 \cdot 7.28 \cdot 10^{-3} = 1.46 \cdot 10^{-2} \textrm{ m} = 1.46 \textrm{ cm}$.

The transverse strain is equal to: 
$e_d = -v e = 0.21 \times 7.28 \cdot 10^{-3} = -1.53 \cdot 10^{-3}$.

After elongation the rod is $\Delta D = e_d \cdot D_0 = -1.53 \cdot 10^{-3} \times 1 \textrm{ cm} = 1.53 \cdot 10^{-3} \textrm{ cm}$ thinner.
`````
