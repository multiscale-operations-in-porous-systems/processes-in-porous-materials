### Question 1 - Porosity and void ratio 1
The most fundamental of all porous material related properties is that of porosity $\phi$. It is defined as the ratio between volume of voids to that of the bulk material:

:::{math}
:label: por-eq-1
\begin{aligned}
\phi &= \frac{V_{void}}{V_{bulk}} \\
V_{bulk} &= V_{pores}+V_{matrix}
\end{aligned}
:::
A related concept is that of the void ratio, which is more common in geotechnical applications.
:::{math}
:label: por-eq-2

e = \frac{V_{void}}{V_{matrix}}

:::

:::{tip} Solution Question 1

Use equations @por-eq-1 and @por-eq-2. For instance, start by writing expressions for $V_{\text{void}}$.

From equation @por-eq-1:
\begin{equation*}
V_{\text{void}} = \phi V_{\text{void}} + \phi V_{\text{matrix}}
\end{equation*}
From equation @por-eq-2:
\begin{equation*}
V_{\text{void}} = e V_{\text{matrix}}
\end{equation*}
Equate both expressions:
\begin{equation*}
\phi V_{\text{void}} + \phi V_{\text{matrix}} = e V_{\text{matrix}}
\end{equation*}
Divide by $V_{\text{matrix}}$:
\begin{equation*}
\phi \frac{V_{\text{void}}}{V_{\text{matrix}}} + \phi = e
\end{equation*}
Note that $\frac{V_{\text{void}}}{V_{\text{matrix}}}$ is the definition of $e$, so:
\begin{equation*}
\phi \cdot e + \phi = e
\end{equation*}
This can be reorganized to:
\begin{equation*}
\phi(1 + e) = e \implies \phi = \frac{e}{1 + e}
\end{equation*}

:::{tip} Solution Question 4

#### Part 1: Porosity ($\phi$)
A Simple Cubic (SC) packing arrangement contains one sphere per unit cell. The radius of each sphere is $R$. 

The length of one side of the cubic unit cell is exactly equal to the diameter of two touching spheres:
$$a = 2R$$

The total volume of the unit cell ($V_{total}$) is:
$$V_{total} = a^3 = (2R)^3 = 8R^3$$

The solid volume inside the unit cell ($V_{solid}$) consists of 8 corner pieces, where each corner represents 1/8 of a sphere. Thus, the total solid volume is equal to the volume of one single sphere:
$$V_{solid} = \frac{4}{3}\pi R^3$$

The porosity ($\phi$) is the ratio of the void volume to the total volume:
$$\phi = \frac{V_{void}}{V_{total}} = \frac{V_{total} - V_{solid}}{V_{total}} = 1 - \frac{V_{solid}}{V_{total}}$$

Substituting the volume expressions:
$$\phi = 1 - \frac{\frac{4}{3}\pi R^3}{8R^3} = 1 - \frac{\pi}{6}$$

Evaluating numerically:
$$\phi \approx 1 - 0.5236 = 0.476$$

#### Part 2: Specific Surface Area ($a_v$)
The specific surface area per unit volume of the **total bed** ($a_v$) is defined as the total interstitial surface area divided by the total volume of the unit cell.

The surface area of the single sphere inside the unit cell is:
$$A_{sphere} = 4\pi R^2$$

Dividing this by the total unit cell volume ($V_{total} = 8R^3$):
$$a_v = \frac{A_{sphere}}{V_{total}} = \frac{4\pi R^2}{8R^3} = \frac{\pi}{2R}$$

*Alternative method using solid phase specific surface area ($a_s$):*
$$a_s = \frac{\text{Surface Area}}{\text{Solid Volume}} = \frac{4\pi R^2}{\frac{4}{3}\pi R^3} = \frac{3}{R}$$
$$a_v = a_s (1 - \phi) = \frac{3}{R} \left(\frac{\pi}{6}\right) = \frac{\pi}{2R}$$

#### Part 3: Tortuosity Factor ($\tau$)
Geometric tortuosity is defined as the ratio of the actual microscopic fluid path length ($L_e$) to the macroscopic straight line length of the porous medium ($L$):
$$\tau = \frac{L_e}{L}$$

In a Simple Cubic lattice model, the open channels run perfectly straight and parallel along the principal coordinate axes. A fluid molecule traveling parallel to a principal axis experiences no geometric deflections or forced deviations around spheres. 

Therefore, the effective path length is identical to the bed length:
$$\text{Because } L_e = L \implies \tau = 1.0$$

#### a) Kozeny-Carman for a porous material with known matrix particle sizes

The Kozeny-Carman equation is

\begin{equation*}
k=\frac{d^{2}}{180}\frac{\phi^{3}}{(1-\phi)^{2}}
\end{equation*}

where:

- $k = \text{ intrinsic permeability } (\mathrm{m^2})$
- $d = \text{ representative grain diameter } (\mathrm{m})$
- $\phi = \text{ porosity}$

##### Step 1: Determine the effective grain size

The particle size distribution in @tab-psd gives:

| Size (mm) | Percent finer (%) |
|-----------|-------------------|
| 0.20 | 35 |
| 0.10 | 10 |

The effective diameter $d_{10}$ corresponds to 10% finer:

\begin{equation*}
d_{10}=0.10~\mathrm{ mm}
\end{equation*}

Convert to metres:

\begin{equation*}
d_{10}=0.10\times 10^{-3}
=1.0\times10^{-4}\ \mathrm{m}
\end{equation*}

##### Step 2: Calculate permeability for $\phi=0.37$

\begin{equation*}
k= \frac{(1.0\times10^{-4})^2}{180} \frac{(0.37)^3}{(1-0.37)^2}
\end{equation*}

\begin{equation*}
k= \frac{1.0\times10^{-8}}{180} \frac{0.050653}{0.3969}
\end{equation*}

\begin{equation*}
k=7.09\times10^{-12}\ \mathrm{m^2}
\end{equation*}

##### Answer

\begin{equation*}
\boxed{k=7.09\times10^{-12}\ \mathrm{m^2}}
\end{equation*}

#### b) Kozeny-Carman for a porous material with known matrix particle sizes with a porosity decrease

##### The new porosity is

\begin{equation*}
\phi_{\text{new}}
=0.37(1-0.25) = 0.2775
\end{equation*}

##### Using the same grain diameter:

\begin{equation*} 
k_{\text{new}} = \frac{(1.0\times10^{-4})^2}{180} \frac{(0.2775)^3}{(1-0.2775)^2}

\begin{equation*}

#### c) Hazen equation for clean sands (a particular porous material!)
The Hazen equation is an empirical relationship for estimating the hydraulic conductivity of clean sands:

\begin{equation*} 
K = C \cdot d_{10}^{2} 
\end{equation*}

where:

$K$ = hydraulic conductivity
$d_{10}$ = effective grain size
$C$ = empirical coefficient
For clean, uniformly graded sands, a commonly used value is

\begin{equation*} 
C = 100 
\end{equation*}

with $d_{10}$ expressed in cm and $K$ obtained in cm/s.

##### Step 1: Determine $d_{10}$
From @tab-psd:
| Size (mm) | Percent finer (%) |
|-----------|-------------------|
| 0.10      | 10                |

Therefore,

\begin{equation*} 
d_{10}=0.10\ \text{mm} 
\end{equation*}

Convert to cm:

\begin{equation*} 
d_{10}=0.01\ \text{cm} 
\end{equation*}

##### Step 2: Calculate hydraulic conductivity
\begin{equation*} 
K = 100(0.01)^2 
\end{equation*}

\begin{equation*} 
K = 0.01\ \text{cm/s} 
\end{equation*}

Convert to SI units:

\begin{equation*} 
K = 1.0\times10^{-4}\ \text{m/s} 
\end{equation*}

##### Step 3: Convert hydraulic conductivity to intrinsic permeability
The relationship between hydraulic conductivity and intrinsic permeability is

\begin{equation*} 
k=\frac{K\mu}{\rho g} 
\end{equation*}

Using properties of water at $20^\circ\text{C}$:

\begin{equation*} 
\mu = 1.002\times10^{-3}\ \text{Pa}\cdot\text{s} 
\end{equation*}

\begin{equation*} 
\rho = 1000\ \text{kg/m}^3 
\end{equation*}

\begin{equation*} 
g = 9.81\ \text{m/s}^2 
\end{equation*}

Substituting:

\begin{equation*} 
k\frac{(1.0\times10^{-4})(1.002\times10^{-3})} {(1000)(9.81)} 
\end{equation*}

\begin{equation*} 
k = 1.02\times10^{-11}\ \text{m}^2 
\end{equation*}

Answer
\begin{equation*} 
\boxed{k = 1.02\times10^{-11}\ \mathrm{m^2}} 
\end{equation*}

This estimate is somewhat larger than the Kozeny-Carman result because Hazen's equation is an empirical correlation developed for clean sands and tends to give relatively high permeability estimates.

### Question X - Hydraulic head 1
As you hopefully remember, the hydraulic head can be viewed as an expression of the Bernoulli equation, which in its turn is a form of conservation of energy. If you would check the dimensions of the Bernoulli equation, you will see that if you divide all terms in the equation by $\rho g$ the terms obtain the dimension of $[L]$. The benefit is that the somewhat abstract concept of energy is now expressed in a tanglible height. For instance, the height of the top of a water column above a chosen reference level. This is called the hydraulic head and its definition for slow flow, such as groundwater flow, is:
```{math}
:label: hh-eq-1

h \equiv \frac{p}{\rho g}+z

```

:::{tip} Solution
:class: dropdown

#### Step 1: Calculate the exit velocity

The volumetric flow rate is

\begin{equation*}
Q = 2~\mathrm{L/s}
= 2\times10^{-3}~\mathrm{m^3/s}
\end{equation*}

The hose diameter is

\begin{equation*}
d = 2~\mathrm{cm}
= 0.02~\mathrm{m}
\end{equation*}

The cross-sectional area is

\begin{equation*}
A = \frac{\pi d^2}{4}
= \frac{\pi(0.02)^2}{4}
= 3.14\times10^{-4}~\mathrm{m^2}
\end{equation*}

The exit velocity is

\begin{equation*}
v = \frac{Q}{A}
= \frac{2\times10^{-3}}{3.14\times10^{-4}}
= 6.37~\mathrm{m/s}
\end{equation*}

#### Step 2: Determine the hydraulic head at the hose outlet

Bernoulli head is

\begin{equation*}
h = z + \frac{p}{\rho g} + \frac{v^2}{2g}
\end{equation*}

At the hose outlet, the water is exposed to the atmosphere, so

\begin{equation*}
\frac{p}{\rho g}=0
\end{equation*}

The elevation is

\begin{equation*}
z = 134.5~\mathrm{m}
\end{equation*}

The velocity head is:

\begin{equation*}
\frac{v^2}{2g} =\frac{(6.37)^2}{2(9.81)}=2.07~\mathrm{m}
\end{equation*}

Therefore,

\begin{equation*}
h = 134.5 + 2.07 = 136.57~\mathrm{m}
\end{equation*}

##### Answer: Hydraulic Head

\begin{equation*}
\boxed{h = 136.57~\mathrm{m}}
\end{equation*}

#### Step 3: Determine the maximum elevation reached by the water

Assuming the hydraulic head remains constant along the trajectory and that the velocity at the highest point is zero,

\begin{equation*}
v = 0
\end{equation*}

and the pressure is atmospheric:

\begin{equation*}
\frac{p}{\rho g}=0
\end{equation*}

Thus,

\begin{equation*}
h = z_{\max}
\end{equation*}

Substituting the previously calculated head:

\begin{equation*}
z_{\max}=136.57~\mathrm{m}
\end{equation*}

##### Answer: Maximum Elevation

The highest elevation reached by the water jet is

\begin{equation*}
\boxed{z_{\max}=136.57~\mathrm{m}}
\end{equation*}

This corresponds to a rise above the hose outlet of

\begin{equation*}
136.57 - 134.5=2.07~\mathrm{m}
\end{equation*}

or

\begin{equation*}
\boxed{2.07~\mathrm{m}}
\end{equation*}

above the end of the hose.

### Question X - Pore size distributions
So far we have considered simple configurations of the matrix. Consequently the distribution of the pores has also been simple. For instance, we considered a unit cell in the Simple Cubic Packing and computed some properties of that porous material. The underlying assumption was that the unit cell described a homogeneous isotropic porous material completely. So if we were to map the distribution of the grains - or more relevant for our studies - the distribution of the pores we would get a very narrow distribution. For actual porous materials that distribution can be quite different. Of course, the distribution of pore sizes provides already much more information than just the bulk porosity or an average pore size. This means we can also attempt to resolve flow at greater details.

Consider the graphs below.

````python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Instellingen voor de figuur
plt.figure(figsize=(11, 6))

# Assen genereren op een logaritmische schaal (omdat brandstofcellen nm-schaal hebben en zandsteen micrometer-schaal)
# Om ze visueel goed te vergelijken op een genormaliseerde as (of effectieve schaal):
x = np.logspace(-3, 2, 2000)

# --- 1. Ideale Simple Cubic Pakking (Theorie) ---
# We centreren de SC pieken puur hypothetisch rond de 1 voor de visuele vergelijking
r_throat = 0.414
r_body = 0.732
y_sc = 0.4 * norm.pdf(x, r_throat, 0.01) + 0.6 * norm.pdf(x, r_body, 0.01)
y_sc = y_sc / max(y_sc) * 90

# --- 2. Natuurlijke Zandsteen (Monomodaal, breed) ---
# Gecentreerd rond grotere poriën met een brede, natuurlijke spreiding
y_sandstone = norm.pdf(np.log10(x), np.log10(2.0), 0.4)
y_sandstone = y_sandstone / max(y_sandstone) * 70

# --- 3. Brandstofcel Katalysatorlaag (Bispedaal / Bimodal) ---
# Twee duidelijke pieken: één op de micro/mesoschaal en één op de macroschaal
y_fuelcell = 0.6 * norm.pdf(np.log10(x), np.log10(0.03), 0.3) + 0.4 * norm.pdf(np.log10(x), np.log10(0.8), 0.3)
y_fuelcell = y_fuelcell / max(y_fuelcell) * 60

# Plots tekenen
plt.plot(x, y_sc, color='darkblue', linewidth=2.5, label='Simple Cubic Model (Discrete)')
plt.fill_between(x, 0, y_sc, color='darkblue', alpha=0.05)

plt.plot(x, y_sandstone, color='crimson', linewidth=2.5, linestyle='--', label='Sandstone rocks (Continuous, Monomodal)')
plt.fill_between(x, 0, y_sandstone, color='crimson', alpha=0.05)

plt.plot(x, y_fuelcell, color='forestgreen', linewidth=2.5, linestyle='-.', label='Fuel cell catalyst (Bimodal)')
plt.fill_between(x, 0, y_fuelcell, color='forestgreen', alpha=0.05)

# Grafiekopmaak met logaritmische X-as om de uitersten goed te tonen
plt.xscale('log')
plt.title('Pore size distribution (PSD): Comparison between 3 different porous materials based on dummy data', fontsize=14, pad=15)
plt.xlabel('Indicative pore radius (Normalized scale)', fontsize=12)
plt.ylabel('Relative volume fraction (%)', fontsize=12)
plt.xlim(0.005, 20.0)
plt.ylim(0, 100)
plt.grid(True, which="both", linestyle=':', alpha=0.5)

# Belangrijke annotaties
plt.annotate('Secundary Macropores \n(Gas transport)', xy=(0.8, 25), xytext=(2.0, 40),
             arrowprops=dict(arrowstyle="->", color='forestgreen'))
plt.annotate('Primary Mesopores \n(Catalysis reaction)', xy=(0.03, 35), xytext=(0.006, 55),
             arrowprops=dict(arrowstyle="->", color='forestgreen'))

plt.legend(fontsize=11, loc='upper right')
plt.show()
````

#### Part 1: Scaling Grain Distributions to Pore Distributions
Porosity and geometric distribution tracking are dimensionless scaling operations. The geometry of a pore throat is bounded strictly by the surrounding grain surfaces. For any rigid, stable packing configuration, the characteristic pore radius ($r$) scales linearly with the characteristic grain radius ($R$):
$$r \propto R$$

If a catalyst layer and a sandstone share the exact same structural packing geometry but the catalyst's grains are 1000 times smaller (3 orders of magnitude), the individual pore channels will also be exactly 1000 times smaller. 

Because the relative variations (sorting/polydispersity) are identical, the shape of the cumulative Pore Size Distribution (PSD) remains completely unchanged. It merely shifts down uniformly along the logarithmic x-axis (pore radius) by 3 orders of magnitude. Porosity itself is a scale-independent ratio and remains identical for both samples.

---

#### Part 2: Bimodal Pore Size Distribution in Fuel Cells
A Polymer Electrolyte Membrane Fuel Cell (PEMFC) catalyst layer requires an engineered, multi-scale pore environment because it facilitates simultaneous, competing transport phenomena:

* **Micro/Mesopores (< 50 nm):** These ultra-small pores maximize the internal specific surface area ($a_v \\propto 1/R$). This provides an enormous density of active catalytic sites (e.g., Platinum) necessary to achieve high electrochemical reaction rates.
* **Macropores (> 100 nm):** Viscous flow and gas diffusion become severely limited in narrow sub-50 nm channels due to wall collisions (Knudsen diffusion) and liquid water clogging (capillary condensation). Macropores act as transport "highways," allowing oxygen/hydrogen reactant gases to quickly enter and liquid water product to be drained away dynamically without flooding the reaction sites.

#### Part 1: Serial vs. Parallel Configurations
Even if Filter A and Filter B contain the exact same volume fractions of large and small pores, their macroscopic fluid conductivities will differ dramatically due to network topology:

* **Filter B (Parallel Setup):** Fluid flows through the large and small pores simultaneously. Because volumetric flow rate through a pore is highly sensitive to radius ($\text{Flow Rate} \propto R^4$ via Hagen-Poiseuille), the continuous large-pore channels act as low-resistance paths. The overall conductivity is dominated by these open paths, yielding a very high macroscale permeability.
* **Filter A (Serial Setup):** Fluid is forced to pass sequentially through a large pore, then a small pore, then a large pore. In a series network, the total fluid transport is throttled by the highest resistance element in the chain—the narrow pore throats. The small pores act as global volumetric bottlenecks, drastically reducing the overall hydraulic conductivity.

---

#### Part 2: Optimization via Pore Network Modeling (PNM)
Pore Network Modeling demonstrates that an engineer can optimize or tune the permeability of a porous material without changing its overall weight or volumetric porosity ($\phi$). This is achieved by manipulating the **spatial topology**:

* **Coordination Number:** Increasing the number of active inter-pore connections (throats) branching out from each pore body. This creates redundant flow paths, ensuring that if one path is blocked or tight, the fluid can bypass it easily.
* **Eliminating Serialization:** Designing the manufacturing process (e.g., 3D printing, sintered gradients, or targeted acid etching) to group larger pores into continuous, uninterrupted flow streams parallel to the pressure gradient while isolating or limiting bottleneck configurations. This minimizes the effective tortuosity factor ($\tau$) and increases macroscale hydraulic conductivity at a fixed density.


:::

#### Step 1: Write the falling-head equation

For a falling-head permeameter,

\begin{equation*}
K=\frac{aL}{At}\ln\left(\frac{h_1}{h_2}\right)
\end{equation*}

Rearranging to solve for time:

\begin{equation*}
t=\frac{aL}{AK}\ln\left(\frac{h_1}{h_2}\right)
\end{equation*}

where:

- $a$ = cross-sectional area of the burette
- $A$ = cross-sectional area of the sample
- $L$ = sample length
- $K$ = hydraulic conductivity
- $h_1$ = initial head difference
- $h_2$ = final head difference

#### Step 2: Calculate the cross-sectional areas

The sample diameter is

\begin{equation*}
D=8~\mathrm{cm}
\end{equation*}

Therefore,

\begin{equation*}
A=\frac{\pi D^2}{4}
=\frac{\pi(8)^2}{4}
=50.27~\mathrm{cm^2}
\end{equation*}

The burette diameter is

\begin{equation*}
d=2~\mathrm{mm}
=0.2~\mathrm{cm}
\end{equation*}

Therefore,

\begin{equation*}
a=\frac{\pi d^2}{4}
=\frac{\pi(0.2)^2}{4}
=0.03142~\mathrm{cm^2}
\end{equation*}

#### Step 3: Substitute the known values

The hydraulic conductivity is

\begin{equation*}
K=10^{-4}~\mathrm{cm/s}
\end{equation*}

The sample length is

\begin{equation*}
L=22~\mathrm{cm}
\end{equation*}

#### Part 1: Intrinsic Permeability Calculation
We are given a parallel Bundle of Tubes model where the total void space ($\phi = 0.30$) is split equally between two pore radii fractions ($f_1 = 0.5$ and $f_2 = 0.5$):
$$R_1 = 10 \times 10^{-6} \text{ m}$$
$$R_2 = 20 \times 10^{-6} \text{ m}$$

The governing equation derived from the Hagen-Poiseuille relationship for intrinsic permeability ($K$) of a parallel bundle of cylinders is:
$$K = \frac{\phi}{8} \sum (f_i R_i^2)$$

Substitute the known values into the summation:
$$\sum (f_i R_i^2) = 0.5 \cdot (10 \times 10^{-6} \text{ m})^2 + 0.5 \cdot (20 \times 10^{-6} \text{ m})^2$$
$$\sum (f_i R_i^2) = 0.5 \cdot (100 \times 10^{-12} \text{ m}^2) + 0.5 \cdot (400 \times 10^{-12} \text{ m}^2)$$
$$\sum (f_i R_i^2) = (50 \times 10^{-12}) + (200 \times 10^{-12}) = 250 \times 10^{-12} \text{ m}^2$$

Now, multiply by the porosity pre-factor:
$$K = \frac{0.30}{8} \cdot (250 \times 10^{-12} \text{ m}^2)$$
$$K = 0.0375 \cdot (250 \times 10^{-12} \text{ m}^2) = 9.375 \times 10^{-12} \text{ m}^2$$

---

#### Part 2: Underlying Assumptions of Kozeny-Carman
The classic Kozeny-Carman equation simplifies highly complex 3D porous structures into an equivalent hydraulic system. The core assumptions required to map a static, volumetric distribution directly to a dynamic permeability constant include:

* **Non-Interacting Parallel Channels:** The system is treated as a continuous bundle of channels with a uniform mean hydraulic radius. It assumes the fluid moves evenly through these channels without crossing over or stalling.
* **Neglecting Network Topology:** It ignores spatial bottlenecks (alternating expansions and contractions) and multi-directional routing loops (connectivity/coordination numbers).
* **Laminar Viscous Flow:** It assumes creeping flow conditions ($\text{Re} < 1$), where Darcy's Law holds valid and inertial effects are non-existent.
* **Isotropic Homogeneity:** The macroscopic properties are assumed to be evenly averaged across the sample continuum.
